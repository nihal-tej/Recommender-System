```python
## As by seeing data in datset we can observe data is from other than english so we should refine into english

import langid
from deep_translator import GoogleTranslator

def convert_to_english(text):
    try:
        lang, _ = langid.classify(text)
        if lang != 'en':
            return GoogleTranslator(source=lang, target='en').translate(text)
        return text
    except:
        return text


```

```python
translate_cols = [
    'title',
    'domain',
    'metadata',
    'description',
    'language',
    'country'
]

for col in translate_cols:
    items[col] = items[col].fillna('').astype(str)
    items[col] = items[col].apply(convert_to_english)
items['combined'] = (
    items['title'] + ' ' +
    items['domain'] + ' ' +
    items['metadata'] + ' ' +
    items['description']
)
```

## As perform basic preprocessing such as 
Removing i)HTML tags
ii)URLS
iii)Punctutations
iv) stopwords
v) making some of the complex english word convertion into full form.

```python
import re

def remove_html_tags(text):
    return re.sub(r'<.*?>', '', text)


def remove_urls(text):
    return re.sub(
        r'https?://\S+|www\.\S+',
        '',
        text
    )


def preprocess(text):
    text = str(text).lower().strip()
    text = remove_html_tags(text)
    text = remove_urls(text)
    text = re.sub(r'\s+', ' ', text)

    return text.strip()
```

```python
for col in translate_cols:
    items[col] = items[col].apply(preprocess)
items['combined'] = (
    "Title: " + items['title'] + " "
    + "Domain: " + items['domain'] + " "
    + "Metadata: " + items['metadata'] + " "
    + "Description: " + items['description']
)
```

```python
from sentence_transformers import SentenceTransformer

# Load SBERT model
model = SentenceTransformer('all-MiniLM-L6-v2')

# Text to embed
corpus = items['combined'].tolist()

# Generate normalized embeddings
arr = model.encode(
    corpus,
    convert_to_tensor=True,
    normalize_embeddings=True
)

print("Embedding shape:", arr.shape)
```

```python
from sentence_transformers.util import cos_sim

scores = cos_sim(query_embedding, arr)[0]

top_k = 5
top_results = scores.topk(k=top_k)
```

```python
time_cap = movie_history['time'].quantile(0.99)

movie_history['time_capped'] = (
    movie_history['time']
    .clip(upper=time_cap)
)
```

```python
movie_history['time_norm'] = (
    movie_history['time_capped'] /
    movie_history['time_capped'].max()
)
```

```python
movie_history['time_norm'] = (
    movie_history['time_norm'].fillna(0)
)
```

```python
movie_history['progress_norm'] = (
    movie_history['progress_percentage']
    .fillna(0) / 100
)
```

Convert action into a numerical signal

```python
action_weight = {
    'started': 0.25,
    'paused': 0.50,
    'stopped': 0.60,
    'completed': 1.00
}
```

```python
movie_history['action_score'] = (
    movie_history['action']
    .map(action_weight)
    .fillna(0)
)
```

```python
movie_history['rating_score'] = (
    movie_history['user_rating'] / 5
)
```

```python
movie_history['engagement_score'] = (
    0.35 * movie_history['progress_norm'] +
    0.25 * movie_history['time_norm'] +
    0.25 * movie_history['action_score'] +
    0.15 * movie_history['rating_score'].fillna(
        movie_history['progress_norm']
    )
)
```

```python
movie_interactions = (
    movie_history
    .groupby(
        ['user_id', 'item_id'],
        as_index=False
    )
    .agg(
        interaction_date=('interaction_date', 'max'),
        engagement_score=('engagement_score', 'max')
    )
)
```

```python
print(
    movie_interactions
    .groupby('user_id')
    .size()
    .describe()
)
```

```python
movie_interactions = movie_interactions.sort_values(
    ['user_id', 'interaction_date']
).reset_index(drop=True)
```

```python
train_list = []
test_list = []

for user_id, group in movie_interactions.groupby('user_id'):

    group = group.sort_values('interaction_date')

    # Keep previous interactions as a DataFrame
    train_list.append(group.iloc[:-1])

    # Keep last interaction as a one-row DataFrame
    test_list.append(group.iloc[[-1]])

# Combine DataFrames
train_interactions = pd.concat(
    train_list,
    ignore_index=True
)

test_interactions = pd.concat(
    test_list,
    ignore_index=True
)
```

```python
print("Train shape:", train_interactions.shape)
print("Test shape:", test_interactions.shape)

print("Train users:", train_interactions['user_id'].nunique())
print("Test users:", test_interactions['user_id'].nunique())
```
