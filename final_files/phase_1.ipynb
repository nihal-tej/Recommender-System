## As perform basic preprocessing such as

iv) stopwords
v) making some of the complex english word convertion into full form.

```python
import re
#by regular expressions
def remove_html_tags(txt):
    pat=re.compile('<.*?>')
    return pat.sub(r'',txt)
def remove_urls(text):
    pattern=re.compile(r'https?://\S+|www\.\S+')
    return pattern.sub(r'',text)
import string,time
ex=string.punctuation
def remove_punc(text):
    for char in ex:
        text=text.replace(char,'')
    return text
```

```python
# test
query1= "Blood & water"
query2="deep learning"
query3="romantic movie"
query4="crime thriller"
query5="guitar music"
query6="history of kings"
query7="world war history military strategy"
query_embedding = model.encode(query1, convert_to_tensor=True)
```

```python
from sentence_transformers.util import cos_sim

scores = cos_sim(query_embedding,arr)[0]

top_k = 5
top_results = scores.topk(k=top_k)
```
